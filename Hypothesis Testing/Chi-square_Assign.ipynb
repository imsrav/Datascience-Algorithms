```python
import numpy as np
import scipy.stats as stats
```

```python
# Creating the contingency table based on the provided data.
observed_data = np.array([
    [50, 70],   # Very Satisfied
    [80, 100],  # Satisfied
    [60, 90],   # Neutral
    [30, 50],   # Unsatisfied
    [20, 50]    # Very Unsatisfied
])

# Define the significance level (alpha)
alpha = 0.05
```

```python
print("Task 1: HypothesIs")
print("H0 (Null Hypothesis): There is NO significant association between the type of smart home device purchased and customer satisfaction (they are independent).")
print("H1 (Alternative Hypothesis): There IS a significant association between the type of smart home device purchased and customer satisfaction (they are dependent).\n")
```

```python
# task 2:compute the chi-square statistic
# chi2_contingency calculates the chi2 stat, p-value, degrees of freedom, and expected frequencies.
# correction=False is used because our table is larger than 2x2.
chi2_stat, p_value, dof, expected_frequencies = stats.chi2_contingency(observed_data, correction=False)
print(f"Calculated Chi-Square Statistic: {chi2_stat:.4f}")
```

```python
print(f"Calculated Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"Degrees of Freedom (dof): {dof}")
print("\nExpected Frequencies Table:")
print(expected_frequencies, "\n")
```

```python
# Task3-Determine critical value.
# stats.chi2.ppf finds the critical value for a given alpha and degrees of freedom.
critical_value = stats.chi2.ppf(1 - alpha, dof)
```

```python
print(f"Critical Value from Chi-Square Distribution: {critical_value:.4f}\n")
```

```python
#task4-decision and conclusion
# We check if our calculated chi2 statistic exceeds our critical threshold
if chi2_stat > critical_value:
    print(f"Decision: REJECT the Null Hypothesis (H0).")
    print(f"Reasoning: The calculated Chi-Square statistic ({chi2_stat:.4f}) is GREATER than the critical value ({critical_value:.4f}).")
    print("Conclusion: There is a statistically significant association between the type of smart home device purchased and customer satisfaction.")
else:
    print(f"Decision: FAIL TO REJECT the Null Hypothesis (H0).")
    print(f"Reasoning: The calculated Chi-Square statistic ({chi2_stat:.4f}) is LESS than or EQUAL to the critical value ({critical_value:.4f}).")
    print("Conclusion: There is no statistically significant association between the type of smart home device purchased and customer satisfaction.")
```
